```python
#PARTH 1

import cv2
import numpy as np
from matplotlib import pyplot as plt
def show_image(title, image, cmap=None):
    plt.figure(figsize=(5, 5))
    plt.title(title)
    if cmap:
        plt.imshow(image, cmap=cmap)
    else:
        plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.show()
#PARTH 2

img1 = cv2.imread('/content/drive/MyDrive/CV image/panda pic.png')
img2 = cv2.imread('/content/drive/MyDrive/CV image/weather project.png')
show_image('Panda', img1)
show_image('waeather project', img2)
```

```python
#Resized
img2=cv2.resize(img2,(300,300))
show_image('RESIZED',img2)

#Cropping
cropped_img=img1[100:300,200:300]
show_image('Cropped',cropped_img)
```

```python


#Addition
# Resize img1 to match img2's dimensions before blending
img1 = cv2.resize(img1, (300, 300))
added_img=cv2.addWeighted(img1, 0.5, img2, 0.5, 0)
show_image("Arthmetic Addition (Bleaning)",added_img)

# Subtraction
subtracted_img = cv2.subtract(img1, img2)
show_image("Arithmetic Subtraction", subtracted_img)

```

```python
#Rotated
ROTATED_IMG=cv2.rotate(img1,cv2.ROTATE_90_CLOCKWISE)
show_image("Rotated",ROTATED_IMG)

ROTATED_IMG=cv2.rotate(img2,cv2.ROTATE_90_CLOCKWISE)
show_image("Rotated",ROTATED_IMG)
```

```python
#PARTH 3


rectangle = np.zeros((img1.shape[0], img1.shape[1]), dtype="uint8")
cv2.rectangle(rectangle, (50, 50), (300, 300), 255, -1)
# Create a circle to demonstrate bitwise ops
circle = np.zeros((img1.shape[0], img1.shape[1]), dtype="uint8")
cv2.circle(circle, (200, 200), 100, 255, -1)
# Bitwise AND (Intersection)
bitwise_and = cv2.bitwise_and(rectangle, circle)
show_image("Bitwise AND", bitwise_and, cmap='gray')
# Bitwise OR (Union)
bitwise_or = cv2.bitwise_or(rectangle, circle)
show_image("Bitwise OR", bitwise_or, cmap='gray')
# Bitwise XOR (Exclusive OR)
bitwise_xor = cv2.bitwise_xor(rectangle, circle)
show_image("Bitwise XOR", bitwise_xor, cmap='gray')
# Bitwise NOT (Inversion)
bitwise_not = cv2.bitwise_not(circle)
show_image("Bitwise NOT", bitwise_not, cmap='gray')
```
